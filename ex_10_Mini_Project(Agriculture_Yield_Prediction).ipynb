```python
import pandas as pd

data = {
    'Rainfall': [800, 900, 1000, 750, 1100, 950, 850, 1200, 700, 1050],
    'Temperature': [28, 27, 26, 30, 25, 27, 29, 24, 31, 26],
    'Fertilizer': [100, 120, 150, 90, 160, 130, 110, 180, 80, 140],
    'Area': [2, 2.5, 3, 1.5, 3.5, 2.8, 2.2, 4, 1.2, 3.2],
    'Yield': [2.8, 3.2, 3.8, 2.4, 4.2, 3.6, 3.0, 4.8, 2.1, 4.0]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("agricultural_yield.csv", index=False)

print("Dataset created successfully!")
print(df)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Import dataset
data = pd.read_csv("agricultural_yield.csv")

# Display dataset
print("Agricultural Yield Dataset:")
print(data.head())

# Select input and output
X = data[['Rainfall', 'Temperature', 'Fertilizer', 'Area']]
Y = data['Yield']

# Split dataset
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X_train, Y_train)

# Predict yield
Y_pred = model.predict(X_test)

# Display results
print("\nActual Yield:")
print(Y_test.values)

print("\nPredicted Yield:")
print(Y_pred.round(2))

# Calculate R2 score
score = r2_score(Y_test, Y_pred)

print("\nR2 Score:", round(score, 2))
```
